# Data Cleaning

```python
'''
1. Handle Missing / Null Values
    a. Fill Null Values
        -> .fillna()
        - Continuous Data
        i. Mean
        ii. Median

        -> Skewness -> .skew()
            -> 0 - 0.5 -> Fill Mean Value
            -> < 0 (Negative Skewness) or > 0.5 (Positive Skewness) -> Fill Median Value

        -> Calculate Mean and Median value of a column.
            -> If both mean and median value are equal or almost equal then fill mean value.
            -> If both mean and median value are not equal has vast difference in values then fill median value.
        
        - Discrete / Categorical
        iii. Mode

        - Continuous / Discrete / Categorical
        iv. Custom Record

        
    b. Drop Null Values
        -> .dropna()
    
2. Remove unwanted data
    -> Remove Whitespace
        -> .strip(), .lstrip(), .rstrip()

    -> Remove Regular Expressions (regex) if needed / Remove unwanted texts
        -> .replace()

3. Change Datatypes
    .astype()

4. Handle Duplicate Records
    -> Find duplicate records
        -> .duplicated()

    -> To remove duplicate records
        -> .drop_duplicates()
'''
```

## Handle Duplicate Records

```python
duplicate_data = df[df.duplicated(subset=['EMPLOYEE_ID'], keep=False)]
duplicate_data
```

### Remove duplicate records

```python
df.drop_duplicates(subset='EMPLOYEE_ID', keep='last', inplace=True)
```

### Checking whitespaces

```python
columns = ['Employee_Name', 'Position', 'State', 'Gender', 'MaritalDesc',
       'CitizenDesc', 'HispanicLatino', 'RaceDesc', 'TermReason',
       'EmploymentStatus', 'Department', 'ManagerName', 'RecruitmentSource',
       'PerformanceScore']

# for col in columns:
#     print(col)
#     print('-'*20)
#     unique_values = hr[col].unique()
#     print(unique_values)
#     print('\n')
```

### Creating function for removing whitespaces

```python
def removeWhitespace(dataframe, columns):
    for col in columns:
        dataframe[col] = dataframe[col].str.strip()
    print('Whitespaces removed.')
```

```python
removeWhitespace(hr, columns)
```

## Replace

```python
'''
# Removing unwanted text, regular expressions (regex)
syntax
---------
-> for text/characters/digits
    - str.replace(find, replace_value)

-> for special characters
    - str.replace(find, replace_value, regex=False/True(default))
'''
```

```python
hr['HispanicLatino'].unique()
```

```python
hr['HispanicLatino'] = hr['HispanicLatino'].str.replace('yes', 'Yes')
```

```python
hr['HispanicLatino'] = hr['HispanicLatino'].str.capitalize()
```

```python
hr['Employee_Name'] = hr['Employee_Name'].str.replace(',', '', regex=False)
```

#### Function to convert datetime

```python
def dateConverter(dataframe, *args):
    for col in args:
        hr[col] = pd.to_datetime(hr[col], format='mixed')
    print('Datetime converted.')
```

```python
dateConverter(hr, 'DOB', 'DateofHire', 'DateofTermination', 'LastPerformanceReview_Date')
```
