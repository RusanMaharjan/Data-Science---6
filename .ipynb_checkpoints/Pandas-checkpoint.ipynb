## Loc and ILoc

```python
'''
- Loc -> Location -> Extract data using alphabetical index
- ILoc -> Integer Location -> Extract data using numeric index
'''
```

```python
products.iloc[1:4, 1:]
```

```python
products.loc['B':'D', 'Brand':]
```

```python
products.iloc[
    [0, 1, 4],
    [0, 2]
]
```

```python
products.loc[
    ['A', 'B', 'E'],
    ['Product Name', 'Price']
]
```
