## Zeros and Ones

```python
# zeros / ones (shape, dtype=float(default)/int)
zero_1d = np.zeros(6, dtype=int)
zero_1d
```

```python
one_1d = np.ones(6, dtype=int)
one_1d
```

```python
zero_2d = np.zeros((5, 6), dtype=int)
zero_2d
```

```python
one_2d = np.ones((5, 6), dtype=int)
one_2d
```

## Arange

```python
# arange(start, stop, step)
for i in range(1, 11):
    print(i)
```

```python
numbers = np.arange(1, 61)
numbers
```
