# Conditional Filtering

```python
import pandas as pd
```

```python
df = pd.read_csv('Dataset/Car_Sales.csv')
df.head()
```

```python
# Find car details whose model year is 2017.
# filter_car = df['year'] == 2017
# df[filter_car]

# select * from cars where year = 2017;
df[df['year'] == 2017][['name', 'year', 'selling_price']]
```

```python
# Find car details whose price is in range of 300000 to 600000.
car_price = df[(df['selling_price'] >= 300000) & (df['selling_price'] <= 600000)]
car_price
```

```python
car_price_between = df[df['selling_price'].between(300000, 600000)]
car_price_between
```

```python
# to find unique values
df['fuel'].unique()
```

```python
# to find number of unique values in a column
df['fuel'].nunique()
```

```python
# Find cars whose emmission fuel type is Petrol or CNG.
car_fuel = df[(df['fuel'] == 'Petrol') | (df['fuel'] == 'CNG')]
car_fuel
```

```python
car_fuel_in = df[df['fuel'].isin(['Petrol', 'CNG'])]
# unique_fuel = car_fuel_in['fuel'].unique()
# unique_fuel
car_fuel_in
```

```python
'''
Find car details which is transmission mode is Automatic, Fuel type is Petrol/Diesel and
which is driven for around 20000 to 35000 range.
'''
car_details = df[(df['transmission'] == 'Automatic') & 
                ((df['fuel'] == 'Petrol') | (df['fuel'] == 'Diesel')) &
                ((df['km_driven'] >= 20000) & (df['km_driven']<= 35000))][['name', 'year']]
car_details
```
