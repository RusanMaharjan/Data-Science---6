```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
server = 'RUSAN'
database = 'BikeStores'

connection_url = f"mssql+pyodbc://@{server}/{database}?driver=ODBC+Driver+17+for+SQL+Server&trusted_connection=yes"
engine = create_engine(connection_url)
```

```python
query = '''
    select
    	sc.customer_id, so.customer_id, sc.first_name, sc.last_name, sc.email, sc.street, so.order_status, so.order_date
    from sales.customers sc
    join sales.orders so
    on sc.customer_id = so.customer_id
    where sc.state = 'TX' and so.order_status = 3;
'''
query_df = pd.read_sql(query, engine)
```
